# 03 Feature Engineering

Purpose: create business meaningful features from raw customer churn data and save them into a feature store.


## What is Feature Engineering?

Feature Engineering is the process of creating new useful columns from existing raw data so that machine learning models can learn better patterns.

In this churn project, we create features related to customer engagement, spending behavior, and complaint behavior.


```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

## Load Raw Training Data


```python
df = pd.read_csv("../data/train_data.csv")

print("Raw Data Shape:", df.shape)
df.head()
```

## Feature 1: Engagement Score

Business idea:

Customers with more transactions and higher session time are more engaged.

Formula:

`engagement_score = total_transactions * avg_session_time`


```python
df["engagement_score"] = (
    df["total_transactions"]
    *
    df["avg_session_time"]
)

df[["total_transactions", "avg_session_time", "engagement_score"]].head()
```

## Feature 2: Spend Per Month

Business idea:

This feature shows customer spending intensity compared to customer lifetime.

Formula:

`spend_per_month = monthly_spend / tenure_months`


```python
df["spend_per_month"] = (
    df["monthly_spend"]
    /
    df["tenure_months"]
)

df[["monthly_spend", "tenure_months", "spend_per_month"]].head()
```

## Feature 3: Complaint Ratio

Business idea:

A customer with many support tickets in a short tenure may be unhappy and more likely to churn.

We add `+1` to tenure to avoid division issues.

Formula:

`complaint_ratio = support_ticket_count / (tenure_months + 1)`


```python
df["complaint_ratio"] = (
    df["support_ticket_count"]
    /
    (df["tenure_months"] + 1)
)

df[["support_ticket_count", "tenure_months", "complaint_ratio"]].head()
```

## Check Engineered Features


```python
df[
    [
        "engagement_score",
        "spend_per_month",
        "complaint_ratio"
    ]
].describe()
```

## Feature Store Dataset

Original data had 13 columns.

We added 3 new engineered features.

Expected final shape: `(50000, 16)`


```python
feature_store_df = df.copy()

print("Feature Store Shape:", feature_store_df.shape)
feature_store_df.head()
```

## Save Feature Store

This CSV will be used by the model training notebook.

Output file:

`../feature_store/customer_features.csv`


```python
feature_store_df.to_csv(
    "../feature_store/customer_features.csv",
    index=False
)

print("Feature Store Created Successfully")
```
